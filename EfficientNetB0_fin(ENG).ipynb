```python
# =========================
# 0) Install packages
# =========================
!pip install -q keras-tuner
!apt-get install -y -qq fonts-nanum
!pip install -q -U tensorflow

# =========================
# 1) Imports & global settings
# =========================
import os, json, unicodedata
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import matplotlib
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, ConfusionMatrixDisplay
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.utils import Sequence
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import ModelCheckpoint, ReduceLROnPlateau, EarlyStopping
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, BatchNormalization, Dropout, Input, Concatenate
from tensorflow.keras.applications import EfficientNetB0
import keras_tuner as kt
from google.colab import drive

# Use default English font for plots
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False

# Mount Google Drive
drive.mount('/content/drive')

# =========================
# 2) Paths & (optional) metadata update
# =========================
base_dir  = '/content/drive/MyDrive/Colab Notebooks/testset'
image_dir = os.path.join(base_dir, 'TS_Color_위해물품')
label_dir = os.path.join(base_dir, 'TL_위해물품')

density_info = {
    "metal_density_high": 8.0, "metal_density_low": 5.0,
    "plastic_density_high": 1.5, "plastic_density_low": 0.8,
    "metal_high_density_color": "#00008B", "metal_low_density_color": "#90EE90",
    "plastic_high_density_color": "#FF8C00", "plastic_low_density_color": "#FFA07A",
}

# Optional JSON meta update
for filename in os.listdir(label_dir):
    if filename.endswith('.json'):
        p = os.path.join(label_dir, filename)
        with open(p, 'r', encoding='utf-8') as f:
            data = json.load(f)
        if "meta" in data:
            for item in data["meta"]:
                if "material_inside" in item:
                    if item["material_inside"] == "Metal":
                        item["density"] = 7.5
                        item["color"] = density_info["metal_high_density_color"] \
                            if item["density"] >= density_info["metal_density_high"] else density_info["metal_low_density_color"]
                    elif item["material_inside"] == "Plastic":
                        item["density"] = 1.2
                        item["color"] = density_info["plastic_high_density_color"] \
                            if item["density"] >= density_info["plastic_density_high"] else density_info["plastic_low_density_color"]
        with open(p, 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=4)
print("All metadata JSON files were updated.")

# =========================
# 3) Collect image paths / labels / (estimated) densities
# =========================
def estimate_density_from_image(image):
    grayscale = tf.image.rgb_to_grayscale(image)
    mean_intensity = tf.reduce_mean(grayscale)
    return (1.0 / (mean_intensity + 1e-5)).numpy()

image_paths, labels, estimated_densities = [], [], []
valid_ext = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff')

class_labels = sorted([d for d in os.listdir(image_dir) if os.path.isdir(os.path.join(image_dir, d))])
class_labels = [unicodedata.normalize('NFC', lb) for lb in class_labels]
num_classes  = len(class_labels)
print(f"Classes ({num_classes}): {class_labels}")

for cl in class_labels:
    cdir = os.path.join(image_dir, cl)
    for name in os.listdir(cdir):
        if name.lower().endswith(valid_ext):
            fp = os.path.join(cdir, name)
            image_paths.append(fp)
            labels.append(unicodedata.normalize('NFC', cl))
            img = tf.io.read_file(fp)
            img = tf.image.decode_image(img, channels=3)
            img = tf.image.resize(img, [224, 224])
            img = tf.cast(img, tf.float32) / 255.0
            estimated_densities.append(estimate_density_from_image(img))

data_df = pd.DataFrame({'filename': image_paths, 'label': labels, 'density': estimated_densities})

scaler = MinMaxScaler()
data_df['density_normalized'] = scaler.fit_transform(data_df[['density']])

label_encoder = LabelEncoder()
data_df['label_encoded'] = label_encoder.fit_transform(data_df['label'])

# =========================
# 4) Split (70/20/10) & class weights
# =========================
train_df, temp_df = train_test_split(data_df, test_size=0.3, random_state=42, stratify=data_df['label'])
val_df,   test_df = train_test_split(temp_df, test_size=0.33, random_state=42, stratify=temp_df['label'])

for df in (train_df, val_df, test_df): df.reset_index(drop=True, inplace=True)

classes = np.unique(train_df['label_encoded'])
cw = compute_class_weight(class_weight='balanced', classes=classes, y=train_df['label_encoded'])
cw = np.clip(cw, 0, 5.0)
class_weight_dict = dict(zip(classes, cw))
print("Class weights:", class_weight_dict)

# =========================
# 5) Generators (image + scalar density)
# =========================
batch_size = 32
datagen_train = ImageDataGenerator(
    rescale=1./255,
    rotation_range=45, width_shift_range=0.3, height_shift_range=0.3,
    shear_range=0.3, zoom_range=0.3, horizontal_flip=True, fill_mode='nearest'
)
datagen_val_test = ImageDataGenerator(rescale=1./255)

class CustomDataGenerator(Sequence):
    def __init__(self, df, batch_size, is_training, datagen):
        super().__init__()
        self.df = df.reset_index(drop=True)
        self.batch_size = batch_size
        self.is_training = is_training
        self.datagen = datagen
        self.indices = np.arange(len(self.df))
        self.on_epoch_end()

    def __len__(self): return int(np.ceil(len(self.df) / self.batch_size))

    def __getitem__(self, idx):
        idxs = self.indices[idx*self.batch_size : (idx+1)*self.batch_size]
        part = self.df.iloc[idxs]

        imgs = []
        for fn in part['filename']:
            im = tf.keras.preprocessing.image.load_img(fn, target_size=(224, 224))
            im = tf.keras.preprocessing.image.img_to_array(im)  # [0..255]
            imgs.append(im)
        imgs = np.array(imgs)
        it = self.datagen.flow(imgs, batch_size=len(imgs), shuffle=False)
        imgs = next(it)  # [0..1]

        dens = part['density_normalized'].values.reshape(-1, 1).astype('float32')
        y    = label_encoder.transform(part['label'])
        y    = tf.keras.utils.to_categorical(y, num_classes=num_classes)

        # return named inputs to align with model Input names
        return {'image_input': imgs, 'density_input': dens}, y

    def on_epoch_end(self):
        if self.is_training: np.random.shuffle(self.indices)

train_generator = CustomDataGenerator(train_df, batch_size, True,  datagen_train)
val_generator   = CustomDataGenerator(val_df,   batch_size, False, datagen_val_test)
test_generator  = CustomDataGenerator(test_df,  batch_size, False, datagen_val_test)

steps_per_epoch  = len(train_generator)
validation_steps = len(val_generator)
test_steps       = len(test_generator)

# =========================
# 6) English class names for all reports/plots
# =========================
class_label_mapping_en = {
    'TS_Color_위해물품_Bullet_총알': 'Bullet',
    'TS_Color_위해물품_Card knife_카드칼(접은 형태)(167)': 'Card knife',
    'TS_Color_위해물품_Electroshock weapon_전자충격기': 'Stun gun',
    'TS_Color_위해물품_Grenade_수류탄': 'Grenade',
    'TS_Color_위해물품_Knife blade_칼날(030)': 'Blade',
    'TS_Color_위해물품_Knife-D_맥가이버 칼(접은 형태)(029)': 'Swiss army knife',
    'TS_Color_위해물품_Knife-E_펜칼': 'Pen knife',
    'TS_Color_위해물품_Pistol_모형 권총(금속 재질)': 'Metal pistol',
    'TS_Color_위해물품_Pistol_모형 권총(플라스틱 재질)': 'Plastic pistol',
    'TS_Color_위해물품_Rifle_모형 소총(플라스틱 재질)': 'Plastic rifle',
    'TS_Color_위해물품_Scissors-A_문구용 가위(접은 형태)(037)': 'Scissors',
}
class_label_mapping_en = {unicodedata.normalize('NFC', k): v for k, v in class_label_mapping_en.items()}
# map index->english name
class_indices = {v: k for k, v in enumerate(label_encoder.classes_)}
inv_class_indices = {v: k for k, v in class_indices.items()}
inv_class_indices_en = {
    k: class_label_mapping_en.get(unicodedata.normalize('NFC', inv_class_indices[k]), inv_class_indices[k])
    for k in inv_class_indices
}
target_names = [inv_class_indices_en[i] for i in range(num_classes)]

# =========================
# 7) Model (EfficientNetB0 + density) with KerasTuner
# =========================
def build_model(hp):
    image_input   = Input(shape=(224, 224, 3), name='image_input')
    density_input = Input(shape=(1,),          name='density_input')

    base_model = EfficientNetB0(input_tensor=image_input, include_top=False, weights='imagenet')
    base_model.trainable = True
    for layer in base_model.layers[:20]:
        layer.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = BatchNormalization()(x)
    x = Concatenate()([x, density_input])

    x = Dense(hp.Int('units_dense1', 256, 1024, step=256), activation='relu')(x)
    x = Dropout(hp.Float('dropout_rate1', 0.3, 0.7, step=0.1))(x)
    x = Dense(hp.Int('units_dense2', 128, 512, step=128), activation='relu')(x)
    x = Dropout(hp.Float('dropout_rate2', 0.3, 0.7, step=0.1))(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=[image_input, density_input], outputs=outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=hp.Float('learning_rate', 1e-5, 1e-3, sampling='log')
        ),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

tuner = kt.Hyperband(
    build_model,
    objective='val_accuracy',
    max_epochs=10,
    factor=3,
    directory='tuner_dir',
    project_name='efficientnetb0_tuning_with_density_en'
)

stop_early = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)

tuner.search(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_generator,
    validation_steps=validation_steps,
    epochs=10,
    callbacks=[stop_early]
)

best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
print(
    "Best hyperparameters:\n"
    f"- units_dense1: {best_hps.get('units_dense1')}\n"
    f"- dropout_rate1: {best_hps.get('dropout_rate1')}\n"
    f"- units_dense2: {best_hps.get('units_dense2')}\n"
    f"- dropout_rate2: {best_hps.get('dropout_rate2')}\n"
    f"- learning_rate: {best_hps.get('learning_rate')}\n"
)

model = tuner.hypermodel.build(best_hps)
model.summary()

# =========================
# 8) Train
# =========================
ckpt_cb = ModelCheckpoint("/content/drive/MyDrive/Colab Notebooks/checkpoints/model_efficientnetb0_with_density_en.h5",
                          save_best_only=True)
rlr_cb  = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)
es_cb   = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

history = model.fit(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=val_generator,
    validation_steps=validation_steps,
    epochs=50,
    callbacks=[ckpt_cb, rlr_cb, es_cb],
    class_weight=class_weight_dict
)

# =========================
# 9) Learning curves (English)
# =========================
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'],     label='Train accuracy')
plt.plot(history.history['val_accuracy'], label='Val accuracy')
plt.title('Training & Validation Accuracy'); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'],     label='Train loss')
plt.plot(history.history['val_loss'], label='Val loss')
plt.title('Training & Validation Loss'); plt.legend()
plt.show()

# =========================
# 10) Evaluation / Predictions / Reports / Confusion Matrix / ROC (English)
# =========================
test_loss, test_acc = model.evaluate(test_generator, steps=test_steps)
print(f"Test Loss: {test_loss:.4f} | Test Accuracy: {test_acc:.4f}")

y_pred_probs   = model.predict(test_generator, steps=test_steps)
y_pred_classes = np.argmax(y_pred_probs, axis=1)

y_true_classes = []
for i in range(test_steps):
    _, by = test_generator[i]
    y_true_classes.extend(np.argmax(by, axis=1))
y_true_classes = np.array(y_true_classes)

# Classification report (English labels)
print(classification_report(y_true_classes, y_pred_classes, target_names=target_names, digits=4))

# Confusion matrix (English)
cm = confusion_matrix(y_true_classes, y_pred_classes)
plt.figure(figsize=(12, 12))
ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=target_names)\
    .plot(cmap=plt.cm.Blues, xticks_rotation=90, values_format='d')
plt.title('Confusion Matrix'); plt.xlabel('Predicted label'); plt.ylabel('True label')
plt.show()

# ROC (English)
from sklearn.preprocessing import label_binarize
y_true_bin = label_binarize(y_true_classes, classes=range(num_classes))
y_pred_probs = y_pred_probs[:len(y_true_bin)]

fpr, tpr, roc_auc = {}, {}, {}
for i in range(num_classes):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

plt.figure(figsize=(10, 8))
for i in range(num_classes):
    plt.plot(fpr[i], tpr[i], label=f'{target_names[i]} (AUC = {roc_auc[i]:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.title('Multi-class ROC'); plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
plt.legend(loc='best'); plt.show()

# =========================
# 11) Correct / Incorrect sample visualization (English captions)
# =========================
# collect images from test generator (dict input → image batch only)
x_test, y_test_cls = [], []
for i in range(test_steps):
    b_inp, b_y = test_generator[i]
    bx = b_inp['image_input']  # [0..1]
    x_test.extend(bx); y_test_cls.extend(np.argmax(b_y, axis=1))
x_test = np.array(x_test); y_test_cls = np.array(y_test_cls)

correct_idx   = np.nonzero(y_pred_classes == y_true_classes)[0]
incorrect_idx = np.nonzero(y_pred_classes != y_true_classes)[0]
pred_names = [target_names[i] for i in y_pred_classes]
true_names = [target_names[i] for i in y_true_classes]

plt.figure(figsize=(10,10))
for i, idx in enumerate(correct_idx[:9]):
    plt.subplot(3,3,i+1); plt.imshow((x_test[idx]*255).astype('uint8'))
    plt.title(f"GT: {true_names[idx]}\nPred: {pred_names[idx]}"); plt.axis('off')
plt.tight_layout(); plt.show()

plt.figure(figsize=(10,10))
for i, idx in enumerate(incorrect_idx[:9]):
    plt.subplot(3,3,i+1); plt.imshow((x_test[idx]*255).astype('uint8'))
    plt.title(f"GT: {true_names[idx]}\nPred: {pred_names[idx]}"); plt.axis('off')
plt.tight_layout(); plt.show()
```
